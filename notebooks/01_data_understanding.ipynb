```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python
print("Unique order IDs:", df_clean['order_id'].nunique())
print("Total rows:", len(df_clean))
print("Duplicate order IDs:", df_clean['order_id'].duplicated().sum())
```

```python
numeric_columns = [
    'age',
    'unit_price_inr',
    'quantity',
    'discount_percent',
    'shipping_fee_inr',
    'gross_amount_inr',
    'discount_amount_inr',
    'net_amount_inr',
    'delivery_days',
    'customer_rating'
]

df_clean[numeric_columns].describe().T
```

```python
df_clean['order_date'] = pd.to_datetime(
    df_clean['order_date'],
    errors='coerce'
)

print("Order date type:", df_clean['order_date'].dtype)
print("Invalid dates:", df_clean['order_date'].isna().sum())
```

```python
categorical_columns = [
    'city',
    'gender',
    'product_category',
    'payment_method',
    'sales_channel',
    'order_status'
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df_clean[col].value_counts(dropna=False))
```

```python
expected_gross = (
    df_clean['unit_price_inr'] * df_clean['quantity']
)

expected_discount = (
    df_clean['gross_amount_inr']
    * df_clean['discount_percent'] / 100
)

expected_net = (
    df_clean['gross_amount_inr']
    - df_clean['discount_amount_inr']
    + df_clean['shipping_fee_inr']
)

print(
    "Gross mismatches:",
    (~np.isclose(df_clean['gross_amount_inr'], expected_gross)).sum()
)

print(
    "Discount mismatches:",
    (~np.isclose(df_clean['discount_amount_inr'], expected_discount)).sum()
)

print(
    "Net amount mismatches:",
    (~np.isclose(df_clean['net_amount_inr'], expected_net)).sum()
)
```

```python
expected_discount = (
    df_clean['gross_amount_inr']
    * df_clean['discount_percent'] / 100
)

difference = (
    df_clean['discount_amount_inr'] - expected_discount
)

mismatch_mask = ~np.isclose(
    df_clean['discount_amount_inr'],
    expected_discount
)

print("Mismatch count:", mismatch_mask.sum())

print("\nFirst 10 mismatched records:")

display(
    df_clean.loc[
        mismatch_mask,
        [
            'gross_amount_inr',
            'discount_percent',
            'discount_amount_inr'
        ]
    ].head(10)
)

print("\nDifference summary:")
print(difference[mismatch_mask].describe())
```

```python
expected_discount = (
    df_clean['gross_amount_inr']
    * df_clean['discount_percent'] / 100
).round(2)

print(
    "Discount mismatches:",
    (~np.isclose(
        df_clean['discount_amount_inr'],
        expected_discount
    )).sum()
)
```

```python
df_clean.to_csv(
    "../data/ecommerce_orders_cleaned.csv",
    index=False
)
```
